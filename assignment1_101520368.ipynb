```
"""
Author: Harrison Pape
Assignment: #1
"""

gym_member = "Alex Alliton"     # string
preferred_weight_kg = 20.5      # float
highest_reps = 25               # int
membership_active = True        # boolean

# dict
workout_stats = {
    "Alex" : (40, 30, 25),
    "Jamie" : (30, 35, 20),
    "Taylor" : (40, 55, 35),
    "Harrison" : (40, 40, 40)
}

# Add new key-value pairs to the workout_stats dictionary
# Add in format of "friendName_Total" : "sum_of_workout_minutes"
for friend, workout_min in list(workout_stats.items()):
    total_min = sum(workout_min)
    workout_stats[f"{friend}_Total"] = total_min

# Creates a 2-D list of each person's workout minutes
# list
workout_list = [list(workout_min) for friend, workout_min in workout_stats.items() if isinstance(workout_min, tuple)]

# Slice off the yoga and running minutes for each person from workout_list into yoga_run
# Iterates through yoga_run and outputs the minutes for those activities with the corresponding person
print(f"\nAll friends yoga and running minutes")
yoga_run = [row[:2] for row in workout_list]
i = 0
for times in yoga_run:
    print(f"{list(workout_stats.keys())[i]}: yoga minutes: {times[0]}, running minutes: {times[1]}")
    i += 1

# Uses negative indexing to output the last two names of the people associated
# with the weightlifting minutes in sliced from workout_list in weight_lift
print(f"\nLast two friends weightlifting minutes")
weight_lift = [row[2] for row in workout_list[-2:]]
j = -2
for weight in weight_lift:
    print(f"{list(workout_stats.keys())[j].replace('_Total', '')}: weightlifting minutes: {weight}")
    j += 1

# Iterate through the workout_stats dictionary
# Outputs a message for people with 120 workout minutes or more
# Uses isinstance function to check only the dictionary entries
# with the total minutes calculated for each person
for friend_total, total_min in workout_stats.items():
    if isinstance(total_min, int) and total_min >= 120:
        friend = friend_total.replace("_Total", "")
        print(f"\nGreat job staying active, {friend}!")

# Get input from the user for a name
name = input("\nEnter a name: ")
# Check if the name is in the workout_stats dictionary
# and print out their corresponding workout info
# Otherwise output the person cannot be found
if name in workout_stats:
    workout_min = workout_stats[name]
    total_min = workout_stats.get(f"{name}_Total")
    print(f"\nStats for {name}:")
    print(f"yoga minutes: {workout_min[0]}")
    print(f"running minutes: {workout_min[1]}")
    print(f"weightlifting minutes: {workout_min[2]}")
    print(f"total minutes: {total_min}")
else:
    print(f"Friend {name} not found in the records.")

# Use the max function to find the friend with the most workout minutes
# uses the lambda function to only compare the total minutes
friend_most, most_min = max(
    ((friend_total, total_min) for friend_total, total_min in workout_stats.items() if isinstance(total_min, int)),
    key = lambda x: x[1])
print(f"\nFriend with most minutes: {friend_most.replace('_Total', '')} with {most_min} minutes")

# Use the min function to find the friend with the least workout minutes
# uses the lambda function to only compare the total minutes
friend_least, least_min = min(
    ((friend_total, total_min) for friend_total, total_min in workout_stats.items() if isinstance(total_min, int)),
    key = lambda x: x[1])
print(f"\nFriend with the least minutes: {friend_least.replace('_Total', '')} with {least_min} minutes")
```
